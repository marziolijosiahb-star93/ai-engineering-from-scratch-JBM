# Exercise 2: load a CSV, show a table, plot a chart

The data is a **made-up** week of salon bookings in `salon_bookings.csv`, just for practice.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
```

## Load the CSV

The last line of a cell is displayed automatically.

```python
df = pd.read_csv("salon_bookings.csv")
df
```

## Add a column

Average revenue per booking (Sunday is closed, so it comes out empty).

```python
df["per_booking"] = (df["revenue"] / df["bookings"]).round(2)
df
```

## Plot it

```python
plt.figure(figsize=(8, 4))
plt.bar(df["day"], df["bookings"])
plt.title("Bookings per day (made-up data)")
plt.ylabel("bookings")
plt.show()
```
