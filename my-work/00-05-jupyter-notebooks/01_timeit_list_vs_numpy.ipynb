# Exercise 1: list comprehension vs NumPy

Make 100,000 random numbers two ways and time each with `%timeit`.

Click a cell and press **Shift+Enter** to run it.

```python
import random

import numpy as np

N = 100_000
```

**Way 1: plain Python.** A loop asks for one random number at a time.

```python
%timeit [random.random() for _ in range(N)]
```

**Way 2: NumPy.** One call makes the whole array at once.

```python
%timeit np.random.rand(N)
```

## What I noticed

Numpy was faster
